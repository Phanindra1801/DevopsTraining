```python
import numpy as np
import pandas as pd

df=pd.read_csv("waste_management.csv")
print(df)
```

```python
print("Dimensions:", df.ndim)
print("Shape:", df.shape)
print("Size:", df.size)

print('Info:',df.info())
print('Describe:',df.describe())
print('Is Null:',df.isnull())
print('Is Null Count:',df.isnull().sum())
```

```python
print("Total waste collected:", df["Daily_Waste_KG"].sum())
print("Maximum waste:", df["Daily_Waste_KG"].max())
print("Minimum waste:", df["Daily_Waste_KG"].min())
print("Average waste:", df["Daily_Waste_KG"].mean())
```

```python
df["Increased_Waste"] = df["Daily_Waste_KG"] * 1.10

df[["Bin_ID", "Daily_Waste_KG", "Increased_Waste"]]
```

```python
high_waste = df[df["Daily_Waste_KG"] > 150]
high_waste
```

```python
high_fill = df[df["Bin_Fill_Percent"] > 80]
high_fill
```

```python
df[(df["Daily_Waste_KG"] > 120) & (df["Recycled_Waste_KG"] < 70)]
```

```python
df["Non_Recycled_Waste_KG"] = (
    df["Daily_Waste_KG"] - df["Recycled_Waste_KG"]
)

df[["Bin_ID", "Daily_Waste_KG",
    "Recycled_Waste_KG", "Non_Recycled_Waste_KG"]]
```

```python
non_recycled = df["Non_Recycled_Waste_KG"]

print("Total non-recycled waste:", non_recycled.sum())
print("Average non-recycled waste:", non_recycled.mean())
print("Minimum non-recycled waste:", non_recycled.min())
print("Maximum non-recycled waste:", non_recycled.max())
```
